```python
# ============================================================
# Cell 1.9.1: Smart install/check TwoSampleMR
# ============================================================

r_code = """
options(repos = c(MRCIEU='https://mrcieu.r-universe.dev', CRAN='https://cloud.r-project.org'))
options(Ncpus = 2)

if (!requireNamespace('TwoSampleMR', quietly = TRUE)) {
  cat('TwoSampleMR is not installed. Installing now...\\n')
  install.packages('TwoSampleMR', dependencies = TRUE)
} else {
  cat('TwoSampleMR is already installed. Skipping installation.\\n')
}

library(TwoSampleMR)
cat('TwoSampleMR loaded successfully.\\n')
cat('Version: ')
print(packageVersion('TwoSampleMR'))
"""

with open("/content/smart_install_twosamplemr.R", "w", encoding="utf-8") as f:
    f.write(r_code)

print("R script written to /content/smart_install_twosamplemr.R")
```

```python
# ============================================================
# Cell 1.11: Check rpy2 connection with R
# ============================================================

import rpy2.robjects as robjects
from rpy2.robjects.packages import importr

# Check whether Python can call R
r_version = robjects.r("R.version.string")[0]
print("R version from rpy2:")
print(r_version)

# Import required R packages through rpy2
TwoSampleMR = importr("TwoSampleMR")
ieugwasr = importr("ieugwasr")
ggplot2 = importr("ggplot2")
data_table = importr("data.table")

print("rpy2 can import TwoSampleMR, ieugwasr, ggplot2, and data.table successfully.")
```

```python
# ============================================================
# Set OpenGWAS JWT token safely
# ============================================================

import os
import getpass
import rpy2.robjects as robjects

OPENGWAS_JWT = getpass.getpass("Paste your OpenGWAS JWT token here: ").strip()

if OPENGWAS_JWT.startswith("Bearer "):
    OPENGWAS_JWT = OPENGWAS_JWT.replace("Bearer ", "").strip()

if len(OPENGWAS_JWT) == 0:
    raise ValueError("OpenGWAS JWT token is empty.")

if not OPENGWAS_JWT.startswith("eyJ"):
    print("WARNING: This token does not start with 'eyJ'. Please check whether you copied the correct JWT token.")

os.environ["OPENGWAS_JWT"] = OPENGWAS_JWT

# Pass token into R session used by rpy2
robjects.r.assign("opengwas_jwt_py", OPENGWAS_JWT)
robjects.r("Sys.setenv(OPENGWAS_JWT = opengwas_jwt_py)")

print("OpenGWAS JWT has been set in Python and R session.")
print("Token length:", len(OPENGWAS_JWT))
print("Token preview:", OPENGWAS_JWT[:10] + "..." + OPENGWAS_JWT[-10:])
```
